```python
import importlib
import sys


def reload_order_orchestration_modules():
    """
    Reloads order_orchestration modules in dependency order,
    followed by the unified execution runner.
    """

    module_names = [
        # Domain models and enums first
        "order_orchestration.domain.enums",
        "order_orchestration.domain.order",
        "order_orchestration.domain.candidate",
        "order_orchestration.domain.decision",

        # State managers
        "order_orchestration.state.inventory_state",
        "order_orchestration.state.git_state",

        # Data utilities
        "order_orchestration.data.moq_resolver",
        "order_orchestration.data.mad_resolver",
        "order_orchestration.data.norms_resolver",
        "order_orchestration.data.rlt_resolver",
        "order_orchestration.data.adapters",

        # Transport and route policy
        "order_orchestration.transport.route_policy",

        # Allocation modules
        "order_orchestration.allocation.allocation_sorting",
        "order_orchestration.allocation.replenishment_git",
        "order_orchestration.allocation.moq_allocator",

        # Workflow modules
        "order_orchestration.workflows.vor",
        "order_orchestration.workflows.push",
        "order_orchestration.workflows.replenishment",

        # Runner last
        "scripts.run_unified_order_execution",
    ]

    reloaded_modules = []

    for module_name in module_names:
        if module_name in sys.modules:
            module = importlib.reload(
                sys.modules[module_name]
            )
        else:
            module = importlib.import_module(
                module_name
            )

        reloaded_modules.append(module_name)

    print("Reloaded modules:")
    for module_name in reloaded_modules:
        print(f"  - {module_name}")

    return sys.modules[
        "scripts.run_unified_order_execution"
    ]


unified_runner = reload_order_orchestration_modules()
```

```python
import pandas as pd


def print_kpi_summary(
    decisions_df: pd.DataFrame,
) -> pd.DataFrame:
    """
    Business KPI summary for daily execution mail.

    Uses unique order line:
        order_id + line_id

    KPI Scope:
        Colour 16
        Colour 17
    """

    if decisions_df is None or decisions_df.empty:
        raise ValueError(
            "decisions_df is empty."
        )

    df = decisions_df.copy()

    # -----------------------------------------------------
    # Normalize
    # -----------------------------------------------------
    for column in [
        "order_id",
        "line_id",
        "decision",
    ]:
        df[column] = (
            df[column]
            .astype("string")
            .str.strip()
            .str.upper()
        )

    df["colour_code"] = pd.to_numeric(
        df["colour_code"],
        errors="coerce",
    ).astype("Int64")

    for column in [
        "source_allocated_quantity",
        "source_git_reserved_quantity",
        "remaining_quantity",
        "requested_quantity",
    ]:
        if column not in df.columns:
            df[column] = 0

        df[column] = pd.to_numeric(
            df[column],
            errors="coerce",
        ).fillna(0)

    # -----------------------------------------------------
    # Credit block flag
    # -----------------------------------------------------
    df["credit_blocked"] = (
        df["decision"]
        == "CREDIT_BLOCKED"
    )


    df["is_allocate_decision"] = (
                df["decision"]
                .eq("ALLOCATE")
            )
    
    # -----------------------------------------------------
    # Aggregate to unique line
    # -----------------------------------------------------
    line_df = (
        df.groupby(
            [
                "order_id",
                "line_id",
            ],
            as_index=False,
            dropna=False,

        )
        
        .agg(
            colour_code=(
                "colour_code",
                "first",
            ),
            credit_blocked=(
                "credit_blocked",
                "max",
            ),
            requested_quantity=(
                "requested_quantity",
                "max",
            ),
            physical_qty=(
                "source_allocated_quantity",
                "sum",
            ),
            git_qty=(
                "source_git_reserved_quantity",
                "sum",
            ),
            remaining_quantity=(
                "remaining_quantity",
                "min",
            ),
            allocated=(
                "is_allocate_decision",
                "max",
            ),
        )
    )

    # -----------------------------------------------------
    # Fulfilled quantity
    # -----------------------------------------------------
    line_df["fulfilled_qty"] = (
        line_df["physical_qty"]
        + line_df["git_qty"]
    )

    # line_df["allocation_done"] = (
    #     line_df["fulfilled_qty"]
    #     > 0
    # )

    line_df["allocation_done"] = (
        line_df["allocated"]
    )
    # -----------------------------------------------------
    # Color buckets
    # -----------------------------------------------------
    color16 = (
        line_df["colour_code"]
        == 16
    )

    color17 = (
        line_df["colour_code"]
        == 17
    )

    # -----------------------------------------------------
    # Counts
    # -----------------------------------------------------
    color16_lines = int(
        color16.sum()
    )

    color17_lines = int(
        color17.sum()
    )

    total_lines = (
        color16_lines
        + color17_lines
    )

    credit_block_lines = int(
        (
            line_df["credit_blocked"]
            & (
                color16
                | color17
            )
        ).sum()
    )

    executable_lines = (
        total_lines
        - credit_block_lines
    )

    # -----------------------------------------------------
    # Executable pools
    # -----------------------------------------------------
    color16_exec = (
        color16
        & ~line_df["credit_blocked"]
    )

    color17_exec = (
        color17
        & ~line_df["credit_blocked"]
    )

    color16_exec_count = int(
        color16_exec.sum()
    )

    color17_exec_count = int(
        color17_exec.sum()
    )

    # -----------------------------------------------------
    # Color 16
    # -----------------------------------------------------
    color16_done = int(
        (
            color16_exec
            & line_df["allocation_done"]
        ).sum()
    )

    color16_not_done = (
        color16_exec_count
        - color16_done
    )

    color16_done_pct = (
        color16_done
        / color16_exec_count
        if color16_exec_count
        else 0
    )

    color16_not_done_pct = (
        color16_not_done
        / color16_exec_count
        if color16_exec_count
        else 0
    )

    # -----------------------------------------------------
    # Color 17
    # -----------------------------------------------------
    color17_done = int(
        (
            color17_exec
            & line_df["allocation_done"]
        ).sum()
    )

    color17_not_done = (
        color17_exec_count
        - color17_done
    )

    color17_done_pct = (
        color17_done
        / color17_exec_count
        if color17_exec_count
        else 0
    )

    color17_not_done_pct = (
        color17_not_done
        / color17_exec_count
        if color17_exec_count
        else 0
    )

    # -----------------------------------------------------
    # Total execution
    # -----------------------------------------------------
    total_done = (
        color16_done
        + color17_done
    )

    execution_pct = (
        total_done
        / executable_lines
        if executable_lines
        else 0
    )

    # -----------------------------------------------------
    # First fill rate
    # -----------------------------------------------------
    total_requested_qty = (
        line_df.loc[
            color16 | color17,
            "requested_quantity",
        ]
        .sum()
    )

    total_fulfilled_qty = (
        line_df.loc[
            color16 | color17,
            "fulfilled_qty",
        ]
        .sum()
    )

    first_fill_rate = (
        total_fulfilled_qty
        / total_requested_qty
        if total_requested_qty
        else 0
    )

    # -----------------------------------------------------
    # Build mail table
    # -----------------------------------------------------
    summary = pd.DataFrame(
        [
            {
                "Color 16 Lines":
                    color16_lines,

                "Color 17 Lines":
                    color17_lines,

                "Total Lines":
                    total_lines,

                "Credit Block Lines":
                    credit_block_lines,

                "Net Total Lines for Execution":
                    executable_lines,

                "Color 16 Allocation Done":
                    (
                        f"{color16_done}/"
                        f"{color16_exec_count}"
                        f" - "
                        f"{color16_done_pct:.0%}"
                    ),

                "Color 16 Allocation Not Done":
                    (
                        f"{color16_not_done}/"
                        f"{color16_exec_count}"
                        f" - "
                        f"{color16_not_done_pct:.0%}"
                    ),

                "Color 17 Allocation Done":
                    (
                        f"{color17_done}/"
                        f"{color17_exec_count}"
                        f" - "
                        f"{color17_done_pct:.0%}"
                    ),

                "Color 17 Allocation Not Done":
                    (
                        f"{color17_not_done}/"
                        f"{color17_exec_count}"
                        f" - "
                        f"{color17_not_done_pct:.0%}"
                    ),

                "Net Total Execution %":
                    (
                        f"{total_done}/"
                        f"{executable_lines}"
                        f" - "
                        f"{execution_pct:.0%}"
                    ), 
            }
        ]
    )

    print()
    print("=" * 80)
    print("DAILY KPI SUMMARY")
    print("=" * 80)
    print(
        summary.to_string(
            index=False
        )
    )
    print()

    return summary
```

```python
import pandas as pd


def print_execution_summary(
    decisions_df: pd.DataFrame,
    parent_rdc_column: str = "parent_plant",
    delivery_days_column: str = "source_sla_days",
) -> None:
    """
    Prints a mail-ready execution summary.

    Execution metrics:
        Calculated at unique order-line level:
            order_id + line_id

    Multi-location line:
        An order line receiving physical/GIT quantity
        from more than one distinct selected_source.

    Parent RDC classification:
        1. Parent RDC only
        2. Other RDC only
        3. Both parent RDC and other RDC

    Delivery-time calculation:
        Calculated at allocation-row level.

        Duplicate order lines are not removed because the same
        order line can be allocated from multiple sources, and
        each source can have a different SLA/delivery time.
    """

    # ---------------------------------------------------------
    # Validate input dataframe
    # ---------------------------------------------------------
    if decisions_df is None or decisions_df.empty:
        print("No decision data available.")
        return

    df = decisions_df.copy()

    required_columns = {
        "order_id",
        "line_id",
        "colour_code",
        "decision",
        "selected_source",
        parent_rdc_column,
    }

    missing_columns = required_columns - set(df.columns)

    if missing_columns:
        raise ValueError(
            "Decision dataframe is missing columns: "
            f"{sorted(missing_columns)}"
        )

    # ---------------------------------------------------------
    # Helper function: normalize plant/RDC values
    #
    # Examples:
    # 1110, 1110.0, "1110", "1110.0" all become "1110"
    # ---------------------------------------------------------
    def normalize_location(
        series: pd.Series,
    ) -> pd.Series:

        numeric_values = pd.to_numeric(
            series,
            errors="coerce",
        )

        normalized_numeric = (
            numeric_values
            .round()
            .astype("Int64")
            .astype("string")
        )

        normalized_string = (
            series
            .astype("string")
            .str.strip()
            .str.upper()
        )

        normalized_result = normalized_numeric.where(
            numeric_values.notna(),
            normalized_string,
        )

        return normalized_result.replace(
            {
                "": pd.NA,
                "NAN": pd.NA,
                "NONE": pd.NA,
                "<NA>": pd.NA,
            }
        )

    # ---------------------------------------------------------
    # Normalize basic fields
    # ---------------------------------------------------------
    for column in [
        "order_id",
        "line_id",
        "decision",
    ]:
        df[column] = (
            df[column]
            .astype("string")
            .str.strip()
            .str.upper()
        )

    # Make credit-block decision values consistent
    df["decision"] = (
        df["decision"]
        .str.replace(
            " ",
            "_",
            regex=False,
        )
        .str.replace(
            "-",
            "_",
            regex=False,
        )
    )

    df["selected_source"] = normalize_location(
        df["selected_source"]
    )

    df[parent_rdc_column] = normalize_location(
        df[parent_rdc_column]
    )

    df["colour_code"] = pd.to_numeric(
        df["colour_code"],
        errors="coerce",
    ).astype("Int64")

    # ---------------------------------------------------------
    # Normalize quantity columns
    # ---------------------------------------------------------
    quantity_columns = [
        "source_allocated_quantity",
        "source_git_reserved_quantity",
        "remaining_quantity",
    ]

    for column in quantity_columns:

        if column not in df.columns:
            df[column] = 0

        df[column] = pd.to_numeric(
            df[column],
            errors="coerce",
        ).fillna(0)

    # ---------------------------------------------------------
    # Normalize delivery-days column
    # ---------------------------------------------------------
    if delivery_days_column in df.columns:
        df[delivery_days_column] = pd.to_numeric(
            df[delivery_days_column],
            errors="coerce",
        )
    else:
        df[delivery_days_column] = pd.NA

    # ---------------------------------------------------------
    # Row-level allocation flags
    # ---------------------------------------------------------
    df["credit_blocked"] = (
        df["decision"].eq("CREDIT_BLOCKED")
        .fillna(False)
    )

    df["has_physical_allocation"] = (
        df["source_allocated_quantity"] > 0
    )

    df["has_git_reservation"] = (
        df["source_git_reserved_quantity"] > 0
    )

    df["is_allocation_row"] = (
        df["has_physical_allocation"]
        | df["has_git_reservation"]
    )

    # ---------------------------------------------------------
    # Parent RDC / other RDC row-level flags
    # ---------------------------------------------------------
    valid_rdc_comparison = (
        df["selected_source"].notna()
        & df[parent_rdc_column].notna()
    )

    df["allocated_from_parent_rdc"] = (
        df["is_allocation_row"]
        & valid_rdc_comparison
        & df["selected_source"].eq(
            df[parent_rdc_column]
        )
    ).fillna(False)

    df["allocated_from_other_rdc"] = (
        df["is_allocation_row"]
        & valid_rdc_comparison
        & df["selected_source"].ne(
            df[parent_rdc_column]
        )
    ).fillna(False)

    df["allocation_source_unclassified"] = (
        df["is_allocation_row"]
        & ~valid_rdc_comparison
    ).fillna(False)

    # ---------------------------------------------------------
    # Aggregate execution metrics to unique order-line level
    # ---------------------------------------------------------
    line_summary = (
        df.groupby(
            [
                "order_id",
                "line_id",
            ],
            as_index=False,
            dropna=False,
        )
        .agg(
            colour_code=(
                "colour_code",
                "first",
            ),
            parent_rdc=(
                parent_rdc_column,
                "first",
            ),
            physical_allocated_quantity=(
                "source_allocated_quantity",
                "sum",
            ),
            git_reserved_quantity=(
                "source_git_reserved_quantity",
                "sum",
            ),
            remaining_quantity=(
                "remaining_quantity",
                "min",
            ),
            credit_blocked=(
                "credit_blocked",
                "max",
            ),
            has_physical_allocation=(
                "has_physical_allocation",
                "max",
            ),
            has_git_reservation=(
                "has_git_reservation",
                "max",
            ),
            allocated_from_parent_rdc=(
                "allocated_from_parent_rdc",
                "max",
            ),
            allocated_from_other_rdc=(
                "allocated_from_other_rdc",
                "max",
            ),
            allocation_source_unclassified=(
                "allocation_source_unclassified",
                "max",
            ),
        )
    )

    # ---------------------------------------------------------
    # Fulfilment calculations
    # ---------------------------------------------------------
    line_summary["total_fulfilled_quantity"] = (
        line_summary["physical_allocated_quantity"]
        + line_summary["git_reserved_quantity"]
    )

    line_summary["fully_allocated"] = (
        ~line_summary["credit_blocked"]
        & (
            line_summary[
                "total_fulfilled_quantity"
            ] > 0
        )
        & (
            line_summary[
                "remaining_quantity"
            ] <= 0
        )
    )

    line_summary["partially_allocated"] = (
        ~line_summary["credit_blocked"]
        & (
            line_summary[
                "total_fulfilled_quantity"
            ] > 0
        )
        & (
            line_summary[
                "remaining_quantity"
            ] > 0
        )
    )

    line_summary["unallocated"] = (
        ~line_summary["credit_blocked"]
        & (
            line_summary[
                "total_fulfilled_quantity"
            ] <= 0
        )
    )

    # ---------------------------------------------------------
    # Mutually exclusive RDC allocation categories
    # ---------------------------------------------------------
    line_summary["allocated_from_parent_only"] = (
        line_summary["allocated_from_parent_rdc"]
        & ~line_summary["allocated_from_other_rdc"]
    )

    line_summary["allocated_from_other_only"] = (
        ~line_summary["allocated_from_parent_rdc"]
        & line_summary["allocated_from_other_rdc"]
    )

    line_summary["allocated_from_parent_and_other"] = (
        line_summary["allocated_from_parent_rdc"]
        & line_summary["allocated_from_other_rdc"]
    )

    # Allocated lines where parent RDC or selected source is missing
    line_summary["allocated_source_unclassified"] = (
        (
            line_summary[
                "total_fulfilled_quantity"
            ] > 0
        )
        & ~line_summary["allocated_from_parent_rdc"]
        & ~line_summary["allocated_from_other_rdc"]
    )

    # ---------------------------------------------------------
    # Multi-location allocation calculation
    # ---------------------------------------------------------
    allocation_rows = df[
        df["is_allocation_row"]
        & df["selected_source"].notna()
    ].copy()

    allocation_split_summary = (
        allocation_rows.groupby(
            [
                "order_id",
                "line_id",
            ],
            as_index=False,
            dropna=False,
        )
        .agg(
            contributing_locations=(
                "selected_source",
                "nunique",
            ),
            allocation_rows_after_split=(
                "selected_source",
                "size",
            ),
        )
    )

    multi_location_lines = (
        allocation_split_summary[
            allocation_split_summary[
                "contributing_locations"
            ] > 1
        ]
        .copy()
    )

    multi_location_unique_lines = len(
        multi_location_lines
    )

    multi_location_rows_after_allocation = int(
        multi_location_lines[
            "allocation_rows_after_split"
        ].sum()
    )

    # ---------------------------------------------------------
    # Final unique-order-line metrics
    # ---------------------------------------------------------
    total_unique_lines = len(
        line_summary
    )

    colour_16_lines = int(
        line_summary["colour_code"]
        .eq(16)
        .sum()
    )

    colour_17_lines = int(
        line_summary["colour_code"]
        .eq(17)
        .sum()
    )

    credit_blocked_lines = int(
        line_summary["credit_blocked"].sum()
    )

    executed_lines = (
        total_unique_lines
        - credit_blocked_lines
    )

    fully_allocated_lines = int(
        line_summary[
            "fully_allocated"
        ].sum()
    )

    partially_allocated_lines = int(
        line_summary[
            "partially_allocated"
        ].sum()
    )

    unallocated_lines = int(
        line_summary[
            "unallocated"
        ].sum()
    )

    physical_allocation_lines = int(
        line_summary[
            "has_physical_allocation"
        ].sum()
    )

    git_reservation_lines = int(
        line_summary[
            "has_git_reservation"
        ].sum()
    )

    total_allocated_lines = int(
        (
            line_summary[
                "total_fulfilled_quantity"
            ] > 0
        ).sum()
    )

    parent_rdc_only_lines = int(
        line_summary[
            "allocated_from_parent_only"
        ].sum()
    )

    other_rdc_only_lines = int(
        line_summary[
            "allocated_from_other_only"
        ].sum()
    )

    parent_and_other_rdc_lines = int(
        line_summary[
            "allocated_from_parent_and_other"
        ].sum()
    )

    unclassified_allocation_lines = int(
        line_summary[
            "allocated_source_unclassified"
        ].sum()
    )

    # ---------------------------------------------------------
    # Delivery-time summary
    #
    # Important:
    # This works directly on allocation rows.
    # It does not remove duplicate order_id + line_id values.
    # ---------------------------------------------------------
    delivery_rows = df[
        df["is_allocation_row"]
    ].copy()

    delivery_rows["delivery_days"] = pd.to_numeric(
        delivery_rows[delivery_days_column],
        errors="coerce",
    )

    # Rows having valid delivery-time/SLA information
    delivery_rows_with_sla = delivery_rows[
        delivery_rows["delivery_days"].notna()
    ].copy()

    if not delivery_rows_with_sla.empty:

        delivery_rows_with_sla["delivery_days"] = (
            delivery_rows_with_sla[
                "delivery_days"
            ]
            .round()
            .astype("Int64")
        )

        # Remove zero or negative delivery days if present
        delivery_rows_with_sla = (
            delivery_rows_with_sla[
                delivery_rows_with_sla[
                    "delivery_days"
                ] > 0
            ]
            .copy()
        )

        delivery_time_table = (
            delivery_rows_with_sla.groupby(
                "delivery_days",
                as_index=False,
                dropna=False,
            )
            .size()
            .rename(
                columns={
                    "size": "Total order lines",
                }
            )
            .sort_values(
                "delivery_days",
                ascending=True,
            )
            .reset_index(drop=True)
        )

        delivery_time_table["Delivery time"] = (
            delivery_time_table[
                "delivery_days"
            ].apply(
                lambda days: (
                    "1 day"
                    if days == 1
                    else f"{days} days"
                )
            )
        )

        delivery_time_table = delivery_time_table[
            [
                "Delivery time",
                "Total order lines",
            ]
        ]

    else:
        delivery_time_table = pd.DataFrame(
            columns=[
                "Delivery time",
                "Total order lines",
            ]
        )

    # Count allocation rows without SLA
    allocated_rows_without_delivery_days = int(
        delivery_rows[
            "delivery_days"
        ].isna().sum()
    )

    # Count rows having zero or negative SLA
    allocated_rows_with_invalid_delivery_days = int(
        (
            delivery_rows[
                "delivery_days"
            ].notna()
            & (
                delivery_rows[
                    "delivery_days"
                ] <= 0
            )
        ).sum()
    )

    total_allocated_rows = len(
        delivery_rows
    )

    total_allocated_rows_with_delivery_days = len(
        delivery_rows_with_sla
    )

    # ---------------------------------------------------------
    # Consistency checks
    # ---------------------------------------------------------
    rdc_classified_total = (
        parent_rdc_only_lines
        + other_rdc_only_lines
        + parent_and_other_rdc_lines
        + unclassified_allocation_lines
    )

    delivery_table_total = (
        int(
            delivery_time_table[
                "Total order lines"
            ].sum()
        )
        if not delivery_time_table.empty
        else 0
    )

    # ---------------------------------------------------------
    # Print main execution summary
    # ---------------------------------------------------------
    print()
    print("SUMMARY")
    print()

    print(
        f"Total unique order lines                       : "
        f"{total_unique_lines:,}"
    )
    print(
        f"Colour 16 unique lines                         : "
        f"{colour_16_lines:,}"
    )
    print(
        f"Colour 17 unique lines                         : "
        f"{colour_17_lines:,}"
    )
    print(
        f"Processed order lines                           : "
        f"{executed_lines:,}"
    )
    print(
        f"Not executed due to credit block               : "
        f"{credit_blocked_lines:,}"
    )
    print(
        f"Fully allocated lines                          : "
        f"{fully_allocated_lines:,}"
    )
    print(
        f"Partially allocated lines                      : "
        f"{partially_allocated_lines:,}"
    )
    print(
        f"Unallocated lines                              : "
        f"{unallocated_lines:,}"
    )
    print(
        f"Lines with physical stock allocation           : "
        f"{physical_allocation_lines:,}"
    )
    print(
        f"Lines with GIT reservation                     : "
        f"{git_reservation_lines:,}"
    )
    print(
        f"Total allocated unique lines                   : "
        f"{total_allocated_lines:,}"
    )

    # ---------------------------------------------------------
    # Print allocation-source breakdown
    # ---------------------------------------------------------
    print()
    print("ALLOCATION SOURCE BREAKDOWN")
    print()

    print(
        f"Allocated from parent RDC only                 : "
        f"{parent_rdc_only_lines:,}"
    )
    print(
        f"Allocated from other RDC only                  : "
        f"{other_rdc_only_lines:,}"
    )
    print(
        f"Allocated from parent and other RDC            : "
        f"{parent_and_other_rdc_lines:,}"
    )

    if unclassified_allocation_lines > 0:
        print(
            f"Allocation source could not be classified      : "
            f"{unclassified_allocation_lines:,}"
        )

    print(
        f"Total allocated unique lines in breakdown      : "
        f"{rdc_classified_total:,}"
    )
    print(
        f"Unique lines allocated from >1 location        : "
        f"{multi_location_unique_lines:,}"
    )
    print(
        f"Rows after multi-location allocation split     : "
        f"{multi_location_rows_after_allocation:,}"
    )

    # ---------------------------------------------------------
    # Print delivery-time table
    # ---------------------------------------------------------
    print()
    print("EXPECTED CUSTOMER DELIVERY TIME")
    print()

    print(
        f"Total allocated rows                           : "
        f"{total_allocated_rows:,}"
    )

    if delivery_time_table.empty:
        print(
            "No delivery-time information is available "
            "for allocated rows."
        )
    else:
        print(
            delivery_time_table.to_string(
                index=False,
            )
        )

    print()
    # print(
    #     f"Total allocated rows                           : "
    #     f"{total_allocated_rows:,}"
    # )
    # print(
    #     f"Allocated rows with valid delivery-time data   : "
    #     f"{total_allocated_rows_with_delivery_days:,}"
    # )

    if allocated_rows_without_delivery_days > 0:
        print(
            f"Allocated rows without delivery-time data      : "
            f"{allocated_rows_without_delivery_days:,}"
        )

    if allocated_rows_with_invalid_delivery_days > 0:
        print(
            f"Allocated rows with zero/negative delivery days: "
            f"{allocated_rows_with_invalid_delivery_days:,}"
        )

    # ---------------------------------------------------------
    # Delivery-table consistency check
    # ---------------------------------------------------------
    if (
        delivery_table_total
        != total_allocated_rows_with_delivery_days
    ):
        print()
        print(
            "Warning: Delivery table total does not match "
            "the valid allocated-row count."
        )
```

Cleaning new Ageing inventory file of different formate, date 31st aug 2026

```python
import pandas as pd

input_file = r"c:\Order Orchestration code\Original_files\Copy of combined sheet.29.08.2026.xlsx"
output_file = r"c:\Order Orchestration code\order_orchestration\data\cleaned\cleaned_ageing_plant_material_level.xlsx"

xls = pd.ExcelFile(input_file)

all_data = []

for sheet_name in xls.sheet_names:

    if sheet_name.strip().lower() == "inventory ageing summary":
        continue

    try:

        df = pd.read_excel(
            input_file,
            sheet_name=sheet_name,
            header=1,
            engine="openpyxl"
        )

        # Clean column names
        df.columns = [str(col).strip() for col in df.columns]

        # Dynamic column identification
        plant_col = next(
            col for col in df.columns
            if str(col).strip().lower() == "plant"
        )

        material_col = next(
            col for col in df.columns
            if "material" in str(col).lower()
        )

        stock_col = next(
            col for col in df.columns
            if "stock" in str(col).lower()
        )

        gt1_col = next(
            col for col in df.columns
            if "> 1" in str(col)
        )

        gt2_col = next(
            col for col in df.columns
            if "> 2" in str(col)
        )

        gt3_col = next(
            col for col in df.columns
            if "> 3" in str(col)
        )

        temp = pd.DataFrame()

        temp["plant"] = df[plant_col]
        temp["material_code"] = df[material_col]

        temp["ageing_stock_qty"] = pd.to_numeric(
            df[stock_col],
            errors="coerce"
        ).fillna(0)

        temp["ageing_stock_gt_1yr"] = (
            pd.to_numeric(df[gt1_col], errors="coerce").fillna(0)
            + pd.to_numeric(df[gt2_col], errors="coerce").fillna(0)
            + pd.to_numeric(df[gt3_col], errors="coerce").fillna(0)
        )

        temp["is_ageing_stock_available"] = (
            temp["ageing_stock_gt_1yr"] > 0
        ).astype(int)

        all_data.append(temp)

        print(f"Processed: {sheet_name}")

    except Exception as e:
        print(f"Skipped {sheet_name}: {e}")

if not all_data:
    raise ValueError("No valid sheets processed.")

final_df = pd.concat(all_data, ignore_index=True)

# Remove invalid/header rows
final_df["plant"] = pd.to_numeric(
    final_df["plant"],
    errors="coerce"
)

# final_df["material_code"] = pd.to_numeric(
#     final_df["material_code"],
#     errors="coerce"
# )

final_df = final_df[
    final_df["plant"].notna()
]

final_df = final_df[
    final_df["material_code"].notna()
]

final_df["plant"] = final_df["plant"].astype(int)
final_df["material_code"] = final_df["material_code"].astype(str).str.strip()

# Aggregate duplicates
final_df = (
    final_df.groupby(
        ["plant", "material_code"],
        as_index=False
    )
    .agg({
        "ageing_stock_qty": "sum",
        "ageing_stock_gt_1yr": "sum"
    })
)

# Create ageing availability flag
final_df["is_ageing_stock_available"] = (
    final_df["ageing_stock_gt_1yr"] > 0
).astype(int)

# Final column order
final_df = final_df[
    [
        "plant",
        "material_code",
        "ageing_stock_qty",
        "ageing_stock_gt_1yr",
        "is_ageing_stock_available"
    ]
]

final_df = final_df[final_df['ageing_stock_qty']> 0]

# Save file
final_df.to_excel(output_file, index=False)

print(f"Saved: {output_file}")
print(f"Rows created: {len(final_df):,}")
```
