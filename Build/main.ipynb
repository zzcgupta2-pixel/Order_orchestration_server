```python
open_order_df = (
    open_order_df[
        (
            open_order_df["order_category"]
            .astype(str)
            .str.strip()
            .str.upper()
            .eq("REPLENISHMENT")
        )
        & (
            pd.to_numeric(
                open_order_df["ean_upc"],
                errors="coerce",
            ).isin([16, 17])
        )
    ]
    .assign(
        raw_order_type_sort=lambda df: (
            df["raw_order_type"]
            .astype(str)
            .str.strip()
            .str.upper()
            .map(
                {
                    "ZFMO": 0,
                    "ZMSO": 1,
                }
            )
            .fillna(99)
        ),
        dealer_norms_sort=lambda df: (
            pd.to_numeric(
                df["dealer_norms"],
                errors="coerce",
            )
        ),
    )
    .sort_values(
        by=[
            "ean_upc",
            "raw_order_type_sort",
            "dealer_norms_sort",
        ],
        ascending=[
            True,   # Colour 16 before 17
            True,   # ZFMO before ZMSO
            False,  # Higher dealer norms first
        ],
        na_position="last",
    )
    .drop(
        columns=[
            "raw_order_type_sort",
            "dealer_norms_sort",
        ]
    )
    .reset_index(drop=True)
    .copy()
)
```

```python
open_order_df["parent_plant"] = open_order_df["parent_plant"].astype("Int64")
open_order_df["dealer_norms"] = open_order_df["dealer_norms"].astype("Int64")
```
